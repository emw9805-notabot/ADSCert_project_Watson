```python
# find tables in the document and return table info
# pdfname: https://www.geeksforgeeks.org/python/get-filename-from-path-without-extension-using-python/

def catTable(document, tableStrategy):
    
    pdf = pymupdf.open(document) 
    filename = os.path.basename(document)
    pdfname = filename.split('.',1)[0]

    tableCount = 0
    tablePages = []
    tableHeaders = []
    tableDFs = []


    for page_idx in range(len(pdf)): 
        page = pdf[page_idx]

        find_tables = page.find_tables(strategy = tableStrategy)
        if find_tables.tables:
            table_count = len(find_tables.tables)
            tableheader = find_tables[0].header.names # assume only 1 table per page
            tableDF = find_tables[0].to_pandas()
            tableCount += table_count
            tablePages.append(page_idx)
            tableHeaders.append(tableheader)
            tableDFs.append(tableDF)

    return filename, pdfname, tableCount, tablePages, tableHeaders, tableDFs
```

```python
# assess the tables found in the catTable function to identify the Category tables and extract unique list of category phrases

def tableAssess(language, tableCount, tablePages, tableHeaders, tableDFs):

    if language == 'PT':
        keywordsCAT = ['Categoria']
        nlp = nlp_pt
    elif language == 'ES':
        keywordsCAT = ['Categoría', 'Categorías']
        nlp = nlp_es
    elif language == 'EN':
        keywordsCAT = ['Category', 'Categories', 'Criteria', 'Criterion']
        nlp = nlp_en
    
    tablePageCurrent = 0
    tablePagePrior = 0
    tablepageCurrentTrue = 0
    uniqueCats = []
    uniqueCatsNoDupes = []
    table_idx = []

    if tableCount == 1: # applicable if only 1 table detected in whole pdf
        dataFrame = tableDFs[0]
        header = tableHeaders[0]
        for keyword in keywordsCAT:
            if keyword.lower() in header[0].lower(): # assumes category col is always the first (LHS) col
                colName = header[0]
                uniqueC = dataFrame[colName].dropna().unique().tolist()
                for messyCat in uniqueC:
                    if messyCat:
                        tidyCat = messyCat.replace('\n',' ').lower()
                        uniqueCatsNoDupes.append(tidyCat)
                        table_idx.append(0)
    elif tableCount > 1:
        for idx in range(len(tableDFs)): # count of tableHeaders and DFs should be the same, based on the page level find_tables object
            tablePageCurrent = tablePages[idx]
            dataFrame = tableDFs[idx]
            header = tableHeaders[idx]
            for keyword in keywordsCAT:
                if keyword.lower() in header[0].lower():
                    colName = header[0]
                    uniqueC = dataFrame[colName].dropna().unique().tolist() # unique list of phrases in left hand col of table
                    tablepageCurrentTrue = tablePageCurrent
                    for messyCat in uniqueC:
                        if messyCat:
                            tidyCat = messyCat.replace('\n',' ').lower()
                            uniqueCats.append(tidyCat)
                            table_idx.append(idx)
                elif header[0] == 'Col0' and tablePageCurrent - tablepageCurrentTrue == 1: # the condition where the last keyword-compliant table processed was one page before the current table 
                    uniqueC = dataFrame['Col0'].dropna().unique().tolist()
                    if uniqueC:
                        for messyCat in uniqueC:
                            if messyCat:
                                tidyCat = messyCat.replace('\n',' ').lower()
                                uniqueCats.append(tidyCat)
                                table_idx.append(idx)
            tablePagePrior = tablePageCurrent
        uniqueCatsNoDupes = list(set(uniqueCats))

    return uniqueCatsNoDupes, table_idx
```

```python
# clean up spurious table content and assess presence of enerygy systems

def tableClean(language, uniqueCatsNoDupes):

    sim_th_en = 0.80
    sim_th_pt = 0.80
    sim_th_es = 0.80
    cleanup_TH_en = 0.65
    cleanup_TH_pt = 0.65
    cleanup_TH_es = 0.65

    if language == 'PT':
        re = renewable_energy_pt
        not_re = not_re_cat_words_pt
        all = all_cat_words_pt
        nlp = nlp_pt
        cleanup_TH = cleanup_TH_pt
        similarity_threshold = sim_th_pt
    elif language == 'ES':
        re = renewable_energy_es
        not_re = not_re_cat_words_es
        all = all_cat_words_es
        nlp = nlp_es
        cleanup_TH = cleanup_TH_es
        similarity_threshold = sim_th_es
    elif language == 'EN':
        re = renewable_energy_en
        not_re = not_re_cat_words_en
        all = all_cat_words_en
        nlp = nlp_en
        cleanup_TH = cleanup_TH_en
        similarity_threshold = sim_th_en

    CatwordsFinal = []
    simScores_cln = {}
    notresimScores = {}
    resimScores = {}
    energy_system_tbl = ""

    if len(uniqueCatsNoDupes) > 0:
        for cat in uniqueCatsNoDupes: # phrases from tables' left hand cols
            docCat = nlp(cat) # run phrase through nlp pipeline
            for wordx in all:
                docx = nlp(wordx)
                simScore_cln = docCat.similarity(docx) # compare each cat phrase to all ref words
                if simScore_cln >= cleanup_TH: 
                    CatwordsFinal.append(cat)
                    simScores_cln[cat + ' ' + wordx] = simScore_cln
                    dfsimScores_cln = pd.DataFrame.from_dict(simScores_cln, 'index')
                    CatwordsFFinal = list(set(CatwordsFinal))
        for phrase in CatwordsFFinal: # only phrases that pass the cleanup TH
            docPhrase = nlp(phrase)
            for wordy in docPhrase:
                if wordy.has_vector:
                    for worda in re:
                        doca = nlp(worda)
                        resimScore = wordy.similarity(doca) # compare each word (token) in each cat phrase to renewable energy ref words
                        if resimScore >= similarity_threshold:
                            resimScores[worda] = resimScore
                            dfresimScores = pd.DataFrame.from_dict(resimScores, 'index')
        if resimScores:                     
            energy_system_tbl = True
        else:
            energy_system_tbl = False
            dfresimScores = pd.DataFrame()
    else:
        CatwordsFFinal = []
        dfsimScores_cln = pd.DataFrame()
        dfresimScores = pd.DataFrame()

    return CatwordsFFinal, dfsimScores_cln, dfresimScores, energy_system_tbl, cleanup_TH_en, cleanup_TH_es, cleanup_TH_pt
```

```python
# to find the Use of Proceeds portion of the document, returns font info for key phrases that typically indicate the start and end of that portion

def fontInfo(language, document):

    results_UOP = []
    results_SEEGP = []
    
    pdf = pymupdf.open(document)
    
    if language == 'PT':
        keywordsUOP = ['Usos dos Recursos', 'Uso de Recursos']
        keywordsSEEGP = ['Selection and Evaluation', 'Process for the Project Evaluation and Selection', 'Project Evaluation and Selection Process', 'Project Selection and Evaluation Process', 'Process for Project Evaluation and Selection', 'Project Selection and Assessment Process', 'Project Selection Criteria', 'Project evaluation & selection', 'Project Selection Process']
        keywordsSEEGP_test = ['Avaliação e Seleção de Ativos', 'Processo de Avaliação e Seleção de Projetos']
    elif language == 'ES':
        keywordsUOP = ['Uso de los Fondos', 'Uso de Fondos', 'Uso de Recursos', 'Destinos de Fondos', 'Uso de los Recursos']
        keywordsSEEGP_test = ['Proceso de Evaluación y Selección de Proyectos', 'Selección y Evaluación de Proyectos', 'Procedimiento Selección', 'Evaluación y selección de proyectos']
    elif language == 'EN':
        keywordsUOP = ['Use of Proceeds', 'Use of Funds', 'Use of the Proceeds']
        keywordsSEEGP = ['Selection and Evaluation', 'Process for the Project Evaluation and Selection', 'Project Evaluation and Selection Process', 'Project Selection and Evaluation Process', 'Process for Project Evaluation and Selection', 'Project Selection and Assessment Process', 'Project Selection Criteria', 'Project evaluation & selection', 'Project Selection Process']
        keywordsSEEGP_test = ['Selection and Evaluation', 'Process for', 'Evaluation and Selection', 'Project Evaluation', 'Assessment Process', 'Selection process', 'Project selection criteria']


    for page_idx in range(len(pdf)): 
        page = pdf[page_idx]
        dict = page.get_text("dict")
        blocks = dict["blocks"] 
        for block in blocks:
            if "lines" in block.keys():
                spans = block['lines']
                for span in spans:
                    data = span['spans']
                    for lines in data:
                        for keyword in keywordsUOP:
                            if keyword.lower() in lines['text'].lower().strip():
                                results_UOP.append((lines['text'], lines['size'], lines['bbox'], page_idx))
                        for keyword in keywordsSEEGP_test:
                            if keyword.lower() in lines['text'].lower().strip():
                                results_SEEGP.append((lines['text'], lines['size'], lines['bbox'], page_idx))

                            

    return results_UOP, results_SEEGP
```

```python
# from the results of the previous function, returns the result index containing the highest font size, assuming this is a section header 

def find_max_font(results_UOP, results_SEEGP):
    
    max_font_size_UOP = 0
    for result_idx_UOP in range(len(results_UOP)):
        resultA = results_UOP[result_idx_UOP]
        if resultA[1] > max_font_size_UOP:
            max_font_size_UOP = resultA[1]
            max_idx_UOP = result_idx_UOP

    max_font_size_SEEGP = 0
    if len(results_SEEGP) == 0:
        max_idx_SEEGP = 0
    else:       
        for result_idx_SEEGP in range(len(results_SEEGP)):
            resultB = results_SEEGP[result_idx_SEEGP]
            if resultB[1] > max_font_size_SEEGP:
                max_font_size_SEEGP = resultB[1]
                max_idx_SEEGP = result_idx_SEEGP

    return max_idx_UOP, max_idx_SEEGP
```

```python
# corrects the UOP portion in the cases where the functions above have returned the table of contents instances of the start and end key phrases 

def tocCheck(results_UOP, results_SEEGP, max_idx_UOP, max_idx_SEEGP):
    
    if len(results_SEEGP) == 0:
        areaUOP = results_UOP[max_idx_UOP]
        areaUOP_page = areaUOP[3]
        areaUOP_coords = areaUOP[2]
        areaUOP_y1 = areaUOP_coords[3]
        areaSEEGP_page = 0
        areaSEEGP_y0 = 0

    else:
        areaUOP = results_UOP[max_idx_UOP]
        areaSEEGP = results_SEEGP[max_idx_SEEGP]
        areaUOP_page = areaUOP[3]
        areaSEEGP_page = areaSEEGP[3]
        areaUOP_coords = areaUOP[2]
        areaSEEGP_coords = areaSEEGP[2]
        areaUOP_y1 = areaUOP_coords[3]
        areaSEEGP_y0 = areaSEEGP_coords[1]

        if areaUOP_page == areaSEEGP_page:
            if 0 < areaSEEGP_y0 - areaUOP_y1 < 40:
                results_UOP_remove = results_UOP.pop(max_idx_UOP)
                results_SEEGP_remove = results_SEEGP.pop(max_idx_SEEGP)

                max_idx_UOP, max_idx_SEEGP = find_max_font(results_UOP, results_SEEGP)

                areaUOP = results_UOP[max_idx_UOP]
                areaSEEGP = results_SEEGP[max_idx_SEEGP]
                areaUOP_page = areaUOP[3]
                areaSEEGP_page = areaSEEGP[3]
                areaUOP_coords = areaUOP[2]
                areaSEEGP_coords = areaSEEGP[2]
                areaUOP_y1 = areaUOP_coords[3]
                areaSEEGP_y0 = areaSEEGP_coords[1]

    return areaUOP_page, areaUOP_y1, areaSEEGP_page, areaSEEGP_y0
```

```python
# discards the UOP end boundary in cases where it comes before the start boundary

def keepSEEGP(areaUOP_page, areaUOP_y1, areaSEEGP_page, areaSEEGP_y0):

    if areaSEEGP_page == areaUOP_page:
        if areaSEEGP_y0 < areaUOP_y1:
            keep_areaSEEGP = False
        elif areaSEEGP_y0 > areaUOP_y1:
            keep_areaSEEGP = True
    elif areaSEEGP_page > areaUOP_page:
        keep_areaSEEGP = True
    elif areaSEEGP_page < areaUOP_page:
        keep_areaSEEGP = False

    return keep_areaSEEGP
```

```python
# page scenarios and extract words when keep_areaSEEGP = True
# function called by another function

def wordsUOP_endPage(document, areaUOP_page, areaUOP_y1, areaSEEGP_page, areaSEEGP_y0):

    pdf = pymupdf.open(document)
    extractUOPwords = []

    # iterate pages
    for page_idx in range(len(pdf)):
        page = pdf[page_idx]

       
# process four page scenarios and extract text as words
    # A: UOP all on a single page
        if page_idx == areaUOP_page and page_idx == areaSEEGP_page:
            for word in page.get_text('words'):
                x0, y0, x1, y1, messyText, *_ = word
                if y0 > areaUOP_y1 and y1 < areaSEEGP_y0:
                    tidyText = messyText.replace('\n',' ').lower()
                    extractUOPwords.append(tidyText)

    # UOP across >1 page
    # B: current page is start page
        elif page_idx == areaUOP_page:
            for word in page.get_text('words'):
                x0, y0, x1, y1, messyText, *_ = word
                if y0 > areaUOP_y1:
                    tidyText = messyText.replace('\n',' ').lower()
                    extractUOPwords.append(tidyText)

    # D: current page is neither start nor end page but is in the UOP area
        elif page_idx > areaUOP_page and page_idx < areaSEEGP_page:
            for word in page.get_text('words'):
                x0, y0, x1, y1, messyText, *_ = word
                tidyText = messyText.replace('\n',' ').lower()
                extractUOPwords.append(tidyText)

    # C: current page is end page
        elif page_idx == areaSEEGP_page:
                for word in page.get_text('words'):
                    x0, y0, x1, y1, messyText, *_ = word
                    if y1 < areaSEEGP_y0:
                        tidyText = messyText.replace('\n',' ').lower()
                        extractUOPwords.append(tidyText)

    return extractUOPwords
```

```python
# iterate pages in the pdf and extract words, preferrably between the UOP and SEEGP sections or else from UOP to end

def wordsUOP(keep_areaSEEGP, document, areaUOP_page, areaUOP_y1, areaSEEGP_page, areaSEEGP_y0):
    # language = language

    pdf = pymupdf.open(document)

    if keep_areaSEEGP == True:
        extractUOPwords = wordsUOP_endPage(document, areaUOP_page, areaUOP_y1, areaSEEGP_page, areaSEEGP_y0)
        UOPwordsNoDupes = list(set(extractUOPwords))

    elif keep_areaSEEGP == False:
        extractUOPwords = []

# iterate pages
        for page_idx in range(len(pdf)):
            page = pdf[page_idx]
       
    # process two page scenarios and extract text as words

        # assume UOP across > 1 page
        # B: current page is start page
            if page_idx == areaUOP_page:
                for word in page.get_text('words'):
                    x0, y0, x1, y1, messyText, *_ = word
                    if y0 > areaUOP_y1:
                        tidyText = messyText.replace('\n',' ').lower()
                        extractUOPwords.append(tidyText)

        # E: current page is after the UOP start page
            elif page_idx > areaUOP_page:
                for word in page.get_text('words'):
                    x0, y0, x1, y1, messyText, *_ = word
                    tidyText = messyText.replace('\n',' ').lower()
                    extractUOPwords.append(tidyText)
    
        UOPwordsNoDupes = list(set(extractUOPwords))
    
    return UOPwordsNoDupes
```

```python
# process words into category word dictionaries where doc.similarity score passes threshold

def UOPwords_to_Catwords(language, UOPwordsNoDupes):

    sim_th_en = 0.80
    sim_th_pt = 0.80
    sim_th_es = 0.80

    if language == 'EN':
        nlp = nlp_en
        re = renewable_energy_en
        ee = energy_efficiency_en
        ppc = pollution_prevention_and_control_en
        esml = environmentally_sustainable_management_of_living_natural_resources_and_land_use_en
        tabc = terrestrial_and_aquatic_biodiversity_conservation_en
        ct = clean_transportation_en
        swwm = sustainable_water_and_wastewater_management_en
        cca = climate_change_adaptation_en
        ce = circular_economy_and_or_ecoefficient_projects_en
        gb = green_buildings_en
        similarity_threshold = sim_th_en
    elif language == 'PT':
        nlp = nlp_pt
        re = renewable_energy_pt
        ee = energy_efficiency_pt
        ppc = pollution_prevention_and_control_pt
        esml = environmentally_sustainable_management_of_living_natural_resources_and_land_use_pt
        tabc = terrestrial_and_aquatic_biodiversity_conservation_pt
        ct = clean_transportation_pt
        swwm = sustainable_water_and_wastewater_management_pt
        cca = climate_change_adaptation_pt
        ce = circular_economy_and_or_ecoefficient_projects_pt
        gb = green_buildings_pt
        similarity_threshold = sim_th_pt
    elif language == 'ES':
        nlp = nlp_es
        re = renewable_energy_es
        ee = energy_efficiency_es
        ppc = pollution_prevention_and_control_es
        esml = environmentally_sustainable_management_of_living_natural_resources_and_land_use_es
        tabc = terrestrial_and_aquatic_biodiversity_conservation_es
        ct = clean_transportation_es
        swwm = sustainable_water_and_wastewater_management_es
        cca = climate_change_adaptation_es
        ce = circular_economy_and_or_ecoefficient_projects_es
        gb = green_buildings_es
        similarity_threshold = sim_th_es

    # outputs
    simsre = {}
    energy_system_uop = ""
    simsee = {}
    simsppc = {}
    simsesml = {}
    simstabc = {}
    simsct = {}
    simsswwm = {}
    simscca = {}
    simsce = {}
    simsgb = {}
    cat_count_uop = 0
    

    for wordb in UOPwordsNoDupes:
        docb = nlp(wordb)
        for token in docb: # expect each docb has only 1 token
            if token.has_vector:
    # renewable_energy
                for worda in re:
                    doca = nlp(worda)
                    sim = doca.similarity(docb)
                    if sim >= similarity_threshold:
                        simsre[doca[0].text + ' ' + docb[0].text] = sim

    # energy_efficiency
                for wordc in ee:
                    docc = nlp(wordc)
                    sim = docc.similarity(docb)
                    if sim >= similarity_threshold:
                        simsee[docc[0].text + ' ' + docb[0].text] = sim
    
    # pollution_prevention_and_control
                for wordd in ppc:
                    docd = nlp(wordd)
                    sim = docd.similarity(docb)
                    if sim >= similarity_threshold:
                        simsppc[docd[0].text + ' ' + docb[0].text] = sim

    # environmentally_sustainable_management_of_living_natural_resources_and_land_use
                for worde in esml:
                    doce = nlp(worde)
                    sim = doce.similarity(docb)
                    if sim >= similarity_threshold:
                        simsesml[doce[0].text + ' ' + docb[0].text] = sim

    # terrestrial_and_aquatic_biodiversity_conservation
                for wordf in tabc:
                    docf = nlp(wordf)
                    sim = docf.similarity(docb)
                    if sim >= similarity_threshold:
                        simstabc[docf[0].text + ' ' + docb[0].text] = sim

    # clean_transportation
                for wordg in ct:
                    docg = nlp(wordg)
                    sim = docg.similarity(docb)
                    if sim >= similarity_threshold:
                        simsct[docg[0].text + ' ' + docb[0].text] = sim

    # sustainable_water_and_wastewater_management
                for wordh in swwm:
                    doch = nlp(wordh)
                    sim = doch.similarity(docb)
                    if sim >= similarity_threshold:
                        simsswwm[doch[0].text + ' ' + docb[0].text] = sim

    # climate_change_adaptation
                for wordi in cca:
                    doci = nlp(wordi)
                    sim = doci.similarity(docb)
                    if sim >= similarity_threshold:
                        simscca[doci[0].text + ' ' + docb[0].text] = sim
    
    # circular_economy_and_or_ecoefficient_projects
                for wordj in ce:
                    docj = nlp(wordj)
                    sim = docj.similarity(docb)
                    if sim >= similarity_threshold:
                        simsce[docj[0].text + ' ' + docb[0].text] = sim
    
    # green_buildings
                for wordk in gb:
                    dock = nlp(wordk)
                    sim = dock.similarity(docb)
                    if sim >= similarity_threshold:
                        simsgb[dock[0].text + ' ' + docb[0].text] = sim

    if simsre:
        energy_system_uop = True
        cat_count_uop +=1
    else:
        energy_system_uop = False
    if simsee:
        cat_count_uop +=1
    if simsppc:
        cat_count_uop +=1
    if simsesml:
        cat_count_uop +=1
    if simstabc:
        cat_count_uop +=1
    if simsct:
        cat_count_uop +=1
    if simsswwm:
        cat_count_uop +=1
    if simscca:
        cat_count_uop +=1
    if simsce:
        cat_count_uop +=1
    if simsgb:
        cat_count_uop +=1

    return simsre, simsee, simsppc, simsesml, simstabc, simsct, simsswwm, simscca, simsce, simsgb, sim_th_en, sim_th_es, sim_th_pt, energy_system_uop, cat_count_uop
```
