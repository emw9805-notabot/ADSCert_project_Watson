```python
import pymupdf
import spacy
from spacy.tokens import Span
nlp = spacy.load('en_core_web_sm')

# using the pymupdf Document class
paras=[]
doc = pymupdf.open('/Users/elisabethwatson/Birkbeck/PGC ADS/Module 3 Project/Datasets/GBTP 260201/The-Principles-Guidance-Handbook-June-2025.pdf')
page = doc.load_page(5)
blocks = page.get_text('blocks')
keyword1 = 'categories'
keyword2 = 'green'
paras.append([text for x0, y0, x1, y1, text, block_no, block_type in blocks if keyword1 in text and keyword2 in text])
para = str(paras[0]).replace('\\n', '')
doc = nlp(para)
for token in doc:
    print(token.text, token.i, token.idx)


```
