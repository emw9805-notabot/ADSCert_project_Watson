```python
# https://pymupdf.readthedocs.io/en/latest/the-basics.html

import pymupdf

doc = pymupdf.open("/Users/elisabethwatson/Birkbeck/PGC ADS/Module 3 Project/Datasets/GBTP 260201/inscope framework reports/in scope Bond reports/bono-verde-ingles.pdf") # open a document
out = open("/Users/elisabethwatson/Birkbeck/PGC ADS/Module 3 Project/Datasets/GBTP 260201/inscope framework reports/in scope Bond reports/bono-verde-ingles-output.txt", "wb") # create a text output
for page in doc: # iterate the document pages
    text = page.get_text().encode("utf8") # get plain text (is in UTF-8)
    out.write(text) # write text of page
    out.write(bytes((12,))) # write page delimiter (form feed 0x0C)
out.close()
```
