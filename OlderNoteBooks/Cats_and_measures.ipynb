```python
cats_and_measures_data = pd.read_csv('Cats_and_measures.csv')
cats_and_measures_df = pd.DataFrame(cats_and_measures_data)
cats_and_measures_df.info()
```

```python
cats_and_measures_df['Indicator'].nunique()
```

```python
cats_and_measures_df.groupby('Nomenclature').count()
```
